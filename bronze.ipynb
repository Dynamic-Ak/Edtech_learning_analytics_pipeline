```python

from pyspark.sql.types import *

# 1. Define paths based on your screenshot
base_path = "Files/raw/edtech"

# 2. Ingest Students
df_students_raw = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{base_path}/students.csv")

# 3. Ingest Courses
df_courses_raw = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{base_path}/courses.csv")

# 4. Ingest Assessments
df_assessments_raw = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{base_path}/assessments.csv")

# 5. Write to Tables section as Delta
df_students_raw.write.mode("overwrite").format("delta").saveAsTable("bronze_students")
df_courses_raw.write.mode("overwrite").format("delta").saveAsTable("bronze_courses")
df_assessments_raw.write.mode("overwrite").format("delta").saveAsTable("bronze_assessments")

# 6. Verify row counts (per project requirements)
print(f"Bronze Students: {spark.table('bronze_students').count()} rows")
print(f"Bronze Courses: {spark.table('bronze_courses').count()} rows")
print(f"Bronze Assessments: {spark.table('bronze_assessments').count()} rows")

# Define the destination paths in the Files/bronze folder
bronze_files_path = "Files/bronze"

# Save as Delta format to the Files/bronze directory
df_students_raw.write.format("delta").mode("overwrite").save(f"{bronze_files_path}/students_delta")
df_courses_raw.write.format("delta").mode("overwrite").save(f"{bronze_files_path}/courses_delta")
df_assessments_raw.write.format("delta").mode("overwrite").save(f"{bronze_files_path}/assessments_delta")

print("Files successfully saved to Files/bronze in Delta format.")
```
